````python
import pandas as pd
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
import warnings

warnings.filterwarnings("ignore")

# Load Dataset
df = pd.read_csv("../lo1_lo2_lo3_supervised_learning/dataset/Cleaned_graduation_dataset.csv")
df.head()
````

###  Step 2 — Clean your dataset once not cleaned and apply all preprocessing steps


````python
df.shape
````

###  Step 2 — Splitting Features and Target Variable
In this step, we separate the dataset into:
- **`X`**: The feature matrix (all independent variables used for PCA and model training).  
- **`y`**: The target vector (dependent variable or class label to be predicted).  

We drop the `"class_label"` column from `df` to form `X`, while storing it separately as `y`.


````python
X = df.drop("class_label", axis=1)
y = df['class_label']
````

###  Step 3 — Standardizing the Data
PCA is sensitive to the scale of data.  
In this step, we use **`StandardScaler`** to normalize all features so that each has:
- Mean = 0  
- Standard deviation = 1  

This ensures that features with larger scales don’t dominate the PCA results.


````python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled
````

###  Step 4 — Splitting Dataset into Training and Testing Sets
We divide our scaled dataset into:
- **Training set (80%)** — used to train the model  
- **Testing set (20%)** — used to evaluate model performance  

We use `random_state=30` to ensure reproducibility (same random split every time).


````python
# DATASET SPLIT

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=30)
````

###  Step 5 — Model Training using Logistic Regression
Here, we use **Logistic Regression**, a supervised classification algorithm.  
- The model learns the relationship between features (`X_train`) and target labels (`y_train`).  
- The `.score()` method calculates accuracy on the **test set** to evaluate model performance.


````python
# MODEL Training

model_before_pca = LogisticRegression()
model_before_pca.fit(X_train, y_train)
model_before_pca.score(X_test, y_test)
````

###  Step 6 — Applying PCA (Principal Component Analysis)
In this step, we perform **dimensionality reduction** using PCA.  
- `PCA(0.95)` means the algorithm will automatically choose the minimum number of components that explain **95% of the total variance** in the data.  
- This helps simplify the dataset while preserving most of the important information.  

We then transform the dataset `X` into the new reduced-dimensional space and check the resulting shape of `X_pca`.


# Use PCA to reduce dimensions

````python
# Make transformation
pca_95 = PCA(0.95)
X_pca_9 = pca_95.fit_transform(X)
X_pca.shape
````

###  Step 7 — Checking Explained Variance Ratio
After performing PCA, it’s important to understand how much **variance (information)** each principal component captures.

- The attribute `pca.explained_variance_ratio_` shows the **percentage contribution** of each principal component to the total variance.
- The sum of these values equals approximately **0.95 (95%)**, since we set `PCA(0.95)`.
- Higher values for the first few components indicate they capture most of the important patterns in the dataset.


````python
# Show important/remaining dimensions contribution (%) after reduction
pca_95.explained_variance_ratio_
````

###  Step 8 — Checking the Number of Principal Components Retained
The attribute `pca.n_components_` tells us **how many principal components** were automatically selected by PCA.

- Since we used `PCA(0.95)`, this value represents the **minimum number of components** needed to retain **95% of the dataset’s total variance**.
- This helps confirm how much dimensionality reduction was achieved compared to the original number of features.


````python
# Total remaining Dimensions after reduction
pca_95
````

###  Step 9 — Displaying the Transformed PCA Dataset
At this stage, we view the **transformed dataset** after applying PCA.

- `X_pca` contains the **new feature set** (principal components) that summarize the original features.
- Each column now represents a **principal component (PC1, PC2, PC3, ...)**, capturing key variance patterns.
- The number of columns equals the value shown earlier by `pca.n_components_`.


````python

X_pca
````

###  Step 10 — Training the Model After PCA Reduction
Now that the dataset has been reduced using PCA, we:
1. **Split** the PCA-transformed data (`X_pca`) into training and testing subsets (80%–20% ratio).
2. **Train** a Logistic Regression model using the lower-dimensional data.
3. **Evaluate** the model’s accuracy using the test set.

This allows us to compare the model’s performance **before and after PCA** to see whether dimensionality reduction improves efficiency or accuracy.


````python
# DATASET SPLIT
X_train_pca_95, X_test_pca_95, y_train, y_test = train_test_split(pca_95, y, test_size=0.2, random_state=30)

# MODEL Training After Applying PCA
model_pca_95 = LogisticRegression()
model_pca_95.fit(X_train_pca_95, y_train)
model_pca_95.score(X_test_pca_95, y_test)
````

###  Step 11 — Reducing Data to 2 Principal Components
In this step, we apply PCA again — this time explicitly setting `n_components=2`.

- This forces the PCA to extract only **two principal components** (PC1 and PC2).  
- These two components capture the **maximum possible variance** in just two dimensions.
- This is especially useful for **visualization**, as it allows us to plot the data in a **2D scatter plot** and observe class separation patterns.


````python
pca_2 = PCA(n_components=2)
X_pca_2 = pca.fit_transform(X)
X_pca_2.shape

````

###  Step 12 — Model Training Using 2D PCA Features
After reducing the dataset to **2 principal components**, we:
1. **Split** the transformed dataset (`X_pca`) into training and testing sets (80% for training, 20% for testing).  
2. **Train** a Logistic Regression model using the 2D PCA-transformed data.  
3. **Evaluate** the model’s performance by computing accuracy on the test set.  

This step helps us assess how well a simple 2D PCA representation preserves class-discriminating information for classification tasks.


````python
# DATASET SPLIT

X_train_pca, X_test_pca, y_train, y_test = train_test_split(X_pca_2, y, test_size=0.2, random_state=30)


# MODEL Training After Applying PCA

model_2pca = LogisticRegression()
model_2pca.fit(X_train_pca, y_train)
model_2pca.score(X_test_pca, y_test)
````

##  Step 13 — Model Comparison, Selection, and Deployment

In this final stage, we:
1. **Compare model performances** — evaluate accuracy (or other metrics) of the Logistic Regression model:
   - **Before PCA** → trained on the full original feature set  
   - **After PCA** → trained on reduced-dimensional data  
2. **Select the best-performing model** — based on accuracy, simplicity, and computational efficiency.  
3. **Save the best model** using `joblib` or `pickle` for deployment and future predictions.  
   - Saving ensures we can easily load the model later without retraining.  


````python
import joblib

#  Calculate model accuracies
acc_before_pca = model_before_pca.score(X_test, y_test)       # Original model
acc_pca_95 = model_pca_95.score(X_test_pca_95, y_test)        # PCA retaining 95% variance
acc_pca_2 = model_2pca.score(X_test_pca, y_test)              # PCA with 2 components

#  Display results
print("Model Performance Summary:")
print(f"1️ Accuracy before PCA:        {acc_before_pca:.4f}")
print(f"2️ Accuracy with PCA (0.95):   {acc_pca_95:.4f}")
print(f"3️ Accuracy with PCA (2 PCs):  {acc_pca_2:.4f}")

#  Select the best model
best_model = None
best_acc = max(acc_before_pca, acc_pca_95, acc_pca_2)

if best_acc == acc_before_pca:
    best_model = model_before_pca
    print("\n Original model (without PCA) selected as best model.")
elif best_acc == acc_pca_95:
    best_model = model_pca_95
    print("\n PCA-based model (95% variance) selected as best model.")
else:
    best_model = model_2pca
    print("\n PCA-based model (2 components) selected as best model.")

#  Save the best model for deployment
joblib.dump(best_model, "best_model.pkl")
print("\n Best model saved successfully as 'best_model.pkl'")
````
